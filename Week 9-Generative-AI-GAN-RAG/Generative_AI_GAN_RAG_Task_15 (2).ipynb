### What is Tokenization?

Tokenization is the process of breaking text into smaller pieces called **tokens**.

Example:
"AI is powerful" → ["AI", "is", "powerful"]

### Tasks:
- Tokenize text using BERT tokenizer  
- Count number of tokens  
- Compare with GPT-style tokenization  

```python
!pip install -q transformers
```

```python
from transformers import AutoTokenizer
```

## Step 1: BERT Tokenization

We will use a BERT tokenizer to split text into tokens.

```python
text = "Generative AI is transforming education."

tokenizer_bert = AutoTokenizer.from_pretrained("bert-base-uncased")

tokens_bert = tokenizer_bert.tokenize(text)

print("BERT Tokens:", tokens_bert)
print("Token Count:", len(tokens_bert))
```

## Step 2: GPT-style Tokenization

Now we simulate GPT-style tokenization using another tokenizer.

```python
tokenizer_gpt = AutoTokenizer.from_pretrained("gpt2")

tokens_gpt = tokenizer_gpt.tokenize(text)

print("GPT Tokens:", tokens_gpt)
print("Token Count:", len(tokens_gpt))
```

## Comparison

### Observations:
- Different models break text differently  
- BERT uses subword tokenization  
- GPT uses byte-pair encoding  

---

### Why it matters:
- Token count affects cost (API usage)  
- Tokenization affects model understanding  

```python
new_text = "AI is changing the future of learning."

print("\n--- BERT ---")
print(tokenizer_bert.tokenize(new_text))

print("\n--- GPT ---")
print(tokenizer_gpt.tokenize(new_text))
```

## Reflection

### Which tokenizer produced more tokens?
Usually, BERT produces more tokens because it splits words into smaller subwords.

---

### Key Learning:
- Tokenization is the first step in NLP  
- Different models process text differently  
- Efficient tokenization improves performance  

---

### Conclusion:
Understanding tokens helps in optimizing prompts and reducing API cost.

## Section 3 — RAG Fundamentals (Day 4)

### Objective:
- Understand Retrieval Augmented Generation (RAG)
- Learn embeddings concept
- Perform basic vector search

---

### What is RAG?

RAG = Retrieval + Generation  

It means:
- First, retrieve relevant information  
- Then, generate an answer using that information  

---

### Why RAG?

- LLMs do not know real-time data  
- They may give incorrect answers (hallucination)  
- RAG provides accurate and updated context  

### Tasks:
- Create sample documents  
- Convert text into embeddings  
- Perform similarity search  

```python
!pip install -q sentence-transformers
```

```python
from sentence_transformers import SentenceTransformer
import numpy as np
```

## Step 1: Create Sample Documents

We define a small dataset to search from.

```python
docs = [
    "Artificial Intelligence is the simulation of human intelligence.",
    "Machine Learning is a subset of AI that learns from data.",
    "Deep Learning uses neural networks for complex tasks.",
    "Natural Language Processing helps machines understand text."
]
```

## Step 2: Convert Text into Embeddings

Embeddings are numerical representations of text.

```python
model_embed = SentenceTransformer("all-MiniLM-L6-v2")

doc_embeddings = model_embed.encode(docs)
```

## Step 3: Perform Vector Search

We find the most relevant document based on a query.

```python
query = "What is AI?"

query_embedding = model_embed.encode([query])

scores = np.dot(doc_embeddings, query_embedding.T)

best_index = np.argmax(scores)

print("Query:", query)
print("Most Relevant Document:", docs[best_index])
```

## Step 4: Generate Answer using Context

Now we combine retrieval with AI generation.

```python
context = docs[best_index]

prompt = f"""
Answer using the context below:

Context: {context}

Question: {query}
"""

response = model.generate_content(prompt)

print(response.text)
```

## Analysis

### What happened?
- We retrieved the most relevant document  
- We used it as context for the AI  
- AI generated a more accurate answer  

---

### Key Concept:
RAG improves accuracy by adding external knowledge.

```python
new_query = "What is machine learning?"

query_embedding = model_embed.encode([new_query])
scores = np.dot(doc_embeddings, query_embedding.T)
best_index = np.argmax(scores)

print("Query:", new_query)
print("Answer Source:", docs[best_index])
```

## Reflection

### Benefits of RAG:
- More accurate answers  
- Uses real data  
- Reduces hallucination  

---

### Conclusion:
RAG combines search + AI to improve response quality.

## Section 4 — Build Mini RAG App (Day 5)

### Objective:
- Build a small Question-Answer system
- Use multiple documents
- Combine search + AI generation

---

### What We Will Build:
A simple system where:
1. User asks a question  
2. System finds relevant documents  
3. AI generates an answer using context  

### Tasks:
- Add multiple documents  
- Create embeddings  
- Build query function  
- Generate answers  

```python
docs_extended = [
    "Artificial Intelligence (AI) refers to machines that can think and learn.",
    "Machine Learning (ML) allows systems to learn from data automatically.",
    "Deep Learning is a type of ML using neural networks.",
    "Natural Language Processing (NLP) helps computers understand human language.",
    "AI is used in healthcare, finance, education, and robotics.",
    "Chatbots use AI to communicate with users.",
    "Self-driving cars use AI to make decisions in real time.",
    "AI improves efficiency and automation in industries."
]
```

```python
doc_embeddings_ext = model_embed.encode(docs_extended)
```

## Step 3: Create Query Function

This function will:
- Take a user question  
- Find top relevant documents  
- Generate answer using AI  

```python
def rag_query(query_text, docs, embeddings):

    # Encode query
    query_embedding = model_embed.encode([query_text])

    # Similarity search
    scores = np.dot(embeddings, query_embedding.T)

    # Get top 2 documents
    top_indices = np.argsort(scores.flatten())[-2:]

    context = "\n".join([docs[i] for i in top_indices])

    # Create prompt
    prompt = f"""
    Answer the question using only the context below:

    Context:
    {context}

    Question: {query_text}
    """

    response = model.generate_content(prompt)

    return response.text
```

## Step 4: Test the System

We will ask questions and get AI-generated answers.

```python
questions = [
    "What is AI?",
    "Where is AI used?",
    "What is machine learning?"
]

for q in questions:
    print("Question:", q)
    print("Answer:", rag_query(q, docs_extended, doc_embeddings_ext))
    print("-" * 50)
```

## Analysis

### What happened?
- System searched relevant documents  
- Used them as context  
- Generated accurate answers  

---

### Key Idea:
This is a simple version of ChatGPT-like systems.

```python
def rag_query(query_text, docs, embeddings):

    import time

    query_embedding = model_embed.encode([query_text])
    scores = np.dot(embeddings, query_embedding.T)

    top_index = np.argmax(scores)
    context = docs[top_index]

    prompt = f"""
    Context:
    {context}

    Question: {query_text}

    Answer briefly:
    """

    for i in range(3):
        try:
            response = model.generate_content(prompt)
            time.sleep(2)
            return response.text
        except Exception as e:
            time.sleep(2)

    return "Error: Try again later"
```

## Section 5 — GAN Fundamentals (Day 6)

### Objective:
- Understand Generative Adversarial Networks (GANs)
- Learn Generator vs Discriminator
- Build a simple GAN structure

---

### What is a GAN?

GAN = Generative Adversarial Network  

It consists of two models:
1. Generator → creates fake data  
2. Discriminator → checks real vs fake  

---

### Simple Idea:
Generator tries to fool the Discriminator  
Discriminator tries to detect fake data  
Both improve over time

## Generator vs Discriminator

### Generator:
- Creates fake data (images, text, etc.)
- Takes random noise as input
- Learns to make realistic outputs

### Discriminator:
- Acts like a judge
- Classifies input as real or fake
- Improves detection over time

```python
!pip install -q torch
```

```python
import torch
import torch.nn as nn
```

```python
class Generator(nn.Module):
    def __init__(self):
        super(Generator, self).__init__()

        self.model = nn.Sequential(
            nn.Linear(10, 16),
            nn.ReLU(),
            nn.Linear(16, 8)
        )

    def forward(self, x):
        return self.model(x)
```

## Step 2: Discriminator Model

This model checks whether data is real or fake.

```python
class Discriminator(nn.Module):
    def __init__(self):
        super(Discriminator, self).__init__()

        self.model = nn.Sequential(
            nn.Linear(8, 16),
            nn.ReLU(),
            nn.Linear(16, 1),
            nn.Sigmoid()
        )

    def forward(self, x):
        return self.model(x)
```

## Step 3: Test GAN Models

We generate fake data and pass it through the discriminator.

```python
# Create models
gen = Generator()
disc = Discriminator()

# Generate random noise
noise = torch.randn(1, 10)

# Generate fake data
fake_data = gen(noise)

# Discriminator decision
decision = disc(fake_data)

print("Fake Data:", fake_data)
print("Discriminator Output:", decision)
```

## Explanation

### Process:
1. Generator creates fake data  
2. Discriminator evaluates it  
3. Output shows probability of being real  

---

### Key Idea:
GANs learn by competition between two networks.

```python
# Generate multiple fake samples
for i in range(3):
    noise = torch.randn(1, 10)
    fake = gen(noise)
    print(f"Sample {i+1}:", fake)
```

### Conclusion:
