# Exploratory Data Analysis (EDA) – Titanic Dataset

## Objective
The goal of this notebook is to explore the Titanic dataset and identify key factors that affected passenger survival.

## Dataset Description
The dataset contains information about passengers such as age, gender, ticket class, fare, and survival status.


```python
# Importing required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Loading the Titanic training dataset
df = pd.read_csv("train.csv")
```

```python
# Displaying first 5 rows of the dataset
df.head()
```

```python
# Checking dataset information
df.info()
```

```python
# Checking dataset shape
df.shape
```

## Missing Values Analysis


```python
# Checking missing values in each column
df.isnull().sum()
```

## Survival Analysis


```python
# Counting survived vs not survived passengers
df['Survived'].value_counts()
```

```python
# Plotting survival count
sns.countplot(x='Survived', data=df)
plt.show()
```

## Gender vs Survival


```python
# Survival based on gender
sns.countplot(x='Sex', hue='Survived', data=df)
plt.show()
```

## Class vs Survival


```python
# Survival based on passenger class
sns.countplot(x='Pclass', hue='Survived', data=df)
plt.show()
```

## Age Distribution


```python
# Distribution of age
sns.histplot(df['Age'], bins=30, kde=True)
plt.show()
```

## Final EDA Summary

After performing Exploratory Data Analysis on the Titanic dataset, the following conclusions were drawn:

- Female passengers had a much higher survival rate than male passengers.
- Passengers traveling in first class had a better chance of survival.
- Age played an important role, with children having higher survival rates.
- The Age and Cabin columns contained many missing values.
- Overall, gender, class, and age were the key factors affecting survival.
