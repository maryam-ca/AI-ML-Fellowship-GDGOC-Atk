## 1. Introduction to Logistic Regression

Logistic Regression is a supervised machine learning algorithm used for binary classification problems.

Unlike Linear Regression, it does not predict continuous values. Instead, it predicts probabilities between 0 and 1.

If the probability is greater than 0.5 → class 1  
If the probability is less than 0.5 → class 0

It uses the Sigmoid function to convert linear output into probability.

## 2. Mathematical Background

### 2.1 Linear Combination

First, we calculate:

z = w1x1 + w2x2 + ... + b

This is similar to linear regression.

### 2.2 Sigmoid Function

σ(z) = 1 / (1 + e^(-z))

The sigmoid function converts any real value into a number between 0 and 1.

### 2.3 Cost Function (Binary Cross Entropy)

J = -1/m Σ [ y log(h) + (1-y) log(1-h) ]

Where:
- y = actual value
- h = predicted probability
- m = number of samples

We minimize this cost using Gradient Descent.

## 3. Import Required Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

## 4. Create Dataset

We create a simple binary classification dataset.

```python
X = np.array([[1], [2], [3], [4], [5], [6]])
y = np.array([0, 0, 0, 1, 1, 1])
```

## 5. Define Sigmoid Function

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

## 6. Initialize Parameters

```python
w = 0
b = 0
learning_rate = 0.01
iterations = 1000
```

## 7. Training using Gradient Descent

```python
m = len(X)

for i in range(iterations):

    z = w * X + b
    h = sigmoid(z)

    dw = (1/m) * np.sum((h - y) * X)
    db = (1/m) * np.sum(h - y)

    w = w - learning_rate * dw
    b = b - learning_rate * db
```

## 8. Prediction Function

```python
def predict(X):
    z = w * X + b
    h = sigmoid(z)
    return np.array([1 if i > 0.5 else 0 for i in h])
```

## 9. Model Evaluation

```python
predictions = predict(X)

print("Predictions:", predictions)
print("Actual:", y)

accuracy = np.mean(predictions == y)
print("Accuracy:", accuracy)
```

## 10. Logistic Regression using Scikit-Learn

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X, y)

sklearn_pred = model.predict(X)
print("Sklearn Predictions:", sklearn_pred)
```
