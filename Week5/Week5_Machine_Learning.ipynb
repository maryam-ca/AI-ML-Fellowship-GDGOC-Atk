# Introduction to Machine Learning

Machine Learning is a branch of Artificial Intelligence that allows computers to learn from data without being explicitly programmed.

Instead of writing rules manually, we give data to the machine and it learns patterns automatically.

## Types of Machine Learning

1. Supervised Learning
   - Uses labeled data
   - Example: House price prediction

2. Unsupervised Learning
   - Uses unlabeled data
   - Example: Customer grouping

3. Reinforcement Learning
   - Learns by reward and punishment
   - Example: Game playing AI

## Real Life Applications

# Linear Regression

Linear Regression is a supervised learning algorithm used to predict continuous values.

It finds the best fitting line between input (X) and output (Y).

Equation:

Y = mX + b

Where:
m = slope
b = intercept

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([1,2,3,4,5])
Y = np.array([2,4,5,4,5])
```

```python
mean_x = np.mean(X)
mean_y = np.mean(Y)

numerator = np.sum((X - mean_x) * (Y - mean_y))
denominator = np.sum((X - mean_x)**2)

m = numerator / denominator
b = mean_y - m * mean_x

print("Slope:", m)
print("Intercept:", b)
```

```python
Y_pred = m * X + b
```

```python
plt.scatter(X,Y)
plt.plot(X,Y_pred,color="red")
plt.title("Linear Regression From Scratch")
plt.show()
```

```python
m = 0
b = 0
learning_rate = 0.01
epochs = 1000
n = len(X)
```

```python
for i in range(epochs):
    Y_pred = m*X + b

    dm = (-2/n) * np.sum(X * (Y - Y_pred))
    db = (-2/n) * np.sum(Y - Y_pred)

    m = m - learning_rate * dm
    b = b - learning_rate * db
```

```python
print("Final slope:", m)
print("Final intercept:", b)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X_new = X.reshape(-1,1)
```

```python
model = LinearRegression()
model.fit(X_new, Y)
```

```python
Y_pred = model.predict(X_new)
```

```python
plt.scatter(X,Y)
plt.plot(X,Y_pred,color="green")
plt.title("Scikit Learn Linear Regression")
plt.show()
```
