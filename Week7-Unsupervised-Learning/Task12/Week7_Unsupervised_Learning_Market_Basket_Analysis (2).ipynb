## Task 1 — K-Means Clustering (From Scratch)

```python
# Generate sample data
X, _ = make_blobs(n_samples=200, centers=3, random_state=42)

plt.scatter(X[:,0], X[:,1])
plt.title("Original Data")
plt.show()
```

```python
# K-Means from scratch
def kmeans(X, k=3, iterations=100):
    centroids = X[np.random.choice(len(X), k, replace=False)]

    for _ in range(iterations):
        clusters = [[] for _ in range(k)]

        for point in X:
            distances = [np.linalg.norm(point - c) for c in centroids]
            cluster_index = np.argmin(distances)
            clusters[cluster_index].append(point)

        new_centroids = [np.mean(cluster, axis=0) for cluster in clusters]

        if np.allclose(centroids, new_centroids):
            break

        centroids = new_centroids

    return centroids, clusters

centroids, clusters = kmeans(X)
```

```python
for cluster in clusters:
    cluster = np.array(cluster)
    plt.scatter(cluster[:,0], cluster[:,1])

plt.title("K-Means Clustering (From Scratch)")
plt.show()
```

## Task 2 — Hierarchical Clustering

```python
hc = AgglomerativeClustering(n_clusters=3)
labels = hc.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.title("Hierarchical Clustering")
plt.show()
```

## Task 3 — Dimensionality Reduction (PCA & t-SNE)

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.scatter(X_pca[:,0], X_pca[:,1])
plt.title("PCA Result")
plt.show()
```

## Task 4 — Association Rule Mining (Apriori)

```python
# Sample market basket dataset
data = {
    'Milk': [1,0,1,1,0],
    'Bread': [1,1,1,0,1],
    'Butter': [0,1,1,1,0],
    'Eggs': [1,1,0,1,1]
}

df = pd.DataFrame(data)
df
```

```python
frequent_items = apriori(df, min_support=0.4, use_colnames=True)
frequent_items
```

```python
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.5)
rules
```

## Task 5 — Market Basket Analysis Project

In this project, we apply Apriori algorithm to find relationships between items.

```python
dataset = [
    ['Milk', 'Bread'],
    ['Bread', 'Butter'],
    ['Milk', 'Bread', 'Butter'],
    ['Milk', 'Eggs'],
    ['Bread', 'Eggs']
]

from mlxtend.preprocessing import TransactionEncoder

te = TransactionEncoder()
te_data = te.fit(dataset).transform(dataset)
df = pd.DataFrame(te_data, columns=te.columns_)

df
```

```python
freq_items = apriori(df, min_support=0.4, use_colnames=True)
rules = association_rules(freq_items, metric="lift", min_threshold=1)

rules.sort_values(by='lift', ascending=False)
```
