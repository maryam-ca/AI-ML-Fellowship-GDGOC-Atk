```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.cluster import AgglomerativeClustering
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules
```

## K-Means Clustering

K-Means is an unsupervised learning algorithm used to group data into clusters.

Steps:
1. Choose number of clusters (K)
2. Assign data points to nearest centroid
3. Update centroid
4. Repeat until convergence


```python
from sklearn.datasets import make_blobs

X, y = make_blobs(n_samples=300, centers=3, random_state=42)

kmeans = KMeans(n_clusters=3)
kmeans.fit(X)

plt.scatter(X[:,0], X[:,1], c=kmeans.labels_)
plt.title("K-Means Clustering")
plt.show()
```

## Hierarchical Clustering

Hierarchical clustering builds clusters step by step by merging or splitting clusters.
It is useful when the number of clusters is unknown.

```python
model = AgglomerativeClustering(n_clusters=3)
labels = model.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=labels)
plt.title("Hierarchical Clustering")
plt.show()
```

## Principal Component Analysis (PCA)

PCA reduces the number of features while keeping maximum variance in the data.

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)

plt.scatter(X_pca[:,0], X_pca[:,1])
plt.title("PCA Visualization")
plt.show()
```

## t-SNE Visualization

t-SNE is used to visualize high dimensional data in 2D or 3D.

```python
tsne = TSNE(n_components=2)
X_tsne = tsne.fit_transform(X)

plt.scatter(X_tsne[:,0], X_tsne[:,1])
plt.title("t-SNE Visualization")
plt.show()
```

# Market Basket Analysis Project

Market Basket Analysis helps identify products that are frequently bought together.

We use Association Rule Mining with the Apriori algorithm.

```python
data = {
    'Milk':[1,0,1,1,0],
    'Bread':[1,1,1,0,0],
    'Butter':[0,1,1,0,1],
    'Beer':[1,0,1,0,1]
}

df = pd.DataFrame(data)
df
```

```python
frequent_items = apriori(df, min_support=0.4, use_colnames=True)

frequent_items
```

```python
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.6)

rules
```
